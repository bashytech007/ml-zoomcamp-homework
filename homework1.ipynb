```python
import pandas as pd
import numpy as np

# Q1
print("Q1", pd.__version__)

# Load data
df = pd.read_csv("https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv")

# Q2
print("Q2", len(df))

# Q3
print("Q3", df["fuel_type"].nunique())

# Q4
print("Q4", (df.isnull().sum() > 0).sum())

# Q5
print("Q5", df[df["origin"] == "Asia"]["fuel_efficiency_mpg"].max())

# Q6
med_before = df["horsepower"].median()
mode = df["horsepower"].mode()[0]
med_after = df["horsepower"].fillna(mode).median()
print("Q6", med_before, mode, med_after)

# Q7
X = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]].head(7).values
XTX = X.T @ X
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = np.linalg.inv(XTX) @ X.T @ y
print("Q7", w.sum())
```
